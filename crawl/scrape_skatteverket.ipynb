## Scrape skatteverket.se
- Save whole page and pick out chapter 12 using regex
- sources: 
    - https://docs.firecrawl.dev/features/scrape#scrape
    - Kokchuns lecture
- Help from clude with this part:
```
start = re.search(r"^#+\s*12 kap\.", md, flags=re.MULTILINE)
end = re.search(r"^#+\s*13 kap\.", md, flags=re.MULTILINE)
chapter12 = md[start.start(): end.start() if end else None]
```

````python
import re, os
from dotenv import load_dotenv
from firecrawl import Firecrawl

load_dotenv()
app = Firecrawl(api_key=os.environ["FIRECRAWL_API_KEY"])

url = "https://www4.skatteverket.se/rattsligvagledning/edition/2026.14/27071.html"  # without #-fragment

# Check exact parameter i SDK-documentation
doc = app.scrape(url, formats=["markdown"], only_main_content=True)
md = doc.markdown

os.makedirs("data/raw", exist_ok=True)
with open("data/raw/27071_full.md", "w", encoding="utf-8") as f:
    f.write(md)

# Cut out chapter 12: from title "12 kap." to "13 kap" 
start = re.search(r"^#+\s*12 kap\.", md, flags=re.MULTILINE)
end = re.search(r"^#+\s*13 kap\.", md, flags=re.MULTILINE)
chapter12 = md[start.start(): end.start() if end else None]

with open("data/raw/kap12.md", "w", encoding="utf-8") as f:
    f.write(chapter12)
````
