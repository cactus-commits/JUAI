# Use Firecrawl to crawl chapter 12 from Skatteverket

Source: https://docs.firecrawl.dev/features/crawl

````python
from firecrawl import Firecrawl
from firecrawl.types import ScrapeOptions
import os 
from dotenv import load_dotenv

load_dotenv()

app = Firecrawl(api_key=os.environ["FIRECRAWL_API_KEY"])

url="https://www4.skatteverket.se/rattsligvagledning/edition/2026.14/27071.html#h-12-kap-Vad-som-ska-dras-av-i-inkomstslaget-tjanst"

scrape_opts = ScrapeOptions(
    only_main_content=True,
    max_age=172800000,
    parsers=["pdf"],
    formats=["markdown"]
)

crawl_result = app.crawl(
    url,
    sitemap="include",
    crawl_entire_domain=False,
    limit=10,
    scrape_options=scrape_opts
)

print(crawl_result)
````

## Pick out chapter 12 with regex
Help from clude with:
```
start = re.search(r"^#+\s*12 kap\.", md, flags=re.MULTILINE)
end = re.search(r"^#+\s*13 kap\.", md, flags=re.MULTILINE)
chapter12 = md[start.start(): end.start() if end else None]
```

````python
import re, os
from dotenv import load_dotenv
from firecrawl import Firecrawl

load_dotenv()
app = Firecrawl(api_key=os.environ["FIRECRAWL_API_KEY"])

url = "https://www4.skatteverket.se/rattsligvagledning/edition/2026.14/27071.html" 

doc = app.scrape(url, formats=["markdown"], only_main_content=True)
md = doc.markdown

os.makedirs("data/raw", exist_ok=True)
with open("data/raw/27071_full.md", "w", encoding="utf-8") as f:
    f.write(md)

start = re.search(r"^#+\s*12 kap\.", md, flags=re.MULTILINE)
end = re.search(r"^#+\s*13 kap\.", md, flags=re.MULTILINE)
chapter12 = md[start.start(): end.start() if end else None]

with open("data/raw/kap12.md", "w", encoding="utf-8") as f:
    f.write(chapter12)
````
